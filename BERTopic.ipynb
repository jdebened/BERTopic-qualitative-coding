```python
import os

def preprocess_single_file(file_path):
    data = []
    metadata_headers = ['Newsgroup:', 'Subject:', 'document_id:', 'Document_id:', 'From:']

    with open(file_path, 'r', encoding='utf-8', errors='ignore') as file:
        lines = file.readlines()

    document = []
    for line in lines:
        if any(header in line for header in metadata_headers):
            if document:
                data.append(''.join(document))
                document = []
        else:
            document.append(line)

    if document:
        data.append(''.join(document))

    return data

def preprocess_directory(directory_path):
    data = []
    category_names = []
    categories = []

    files = sorted([f for f in os.listdir(directory_path) if f.endswith('.txt')])

    for group_index, file_name in enumerate(files):
        path = os.path.join(directory_path, file_name)

        file_data = preprocess_single_file(path)
        data.extend(file_data)

        category_names.append(file_name.replace('.txt', ''))
        categories.extend([group_index + 1] * len(file_data))

    return data, categories, category_names

directory_path = './20_newsgroups/'
docs, categories, category_names = preprocess_directory(directory_path)

print(len(docs), len(categories), len(category_names))

```

```python
# Prepare documents
all_docs = docs
doc_chunks = [all_docs[i:i+1000] for i in range(0, len(all_docs), 1000)]
```

```python
from sklearn.cluster import MiniBatchKMeans
from sklearn.decomposition import IncrementalPCA
from bertopic.vectorizers import OnlineCountVectorizer

# Prepare sub-models that support online learning
umap_model = IncrementalPCA(n_components=5)
cluster_model = MiniBatchKMeans(n_clusters=50, random_state=0)
vectorizer_model = OnlineCountVectorizer(stop_words="english", decay=.01)
```

```python
topic_model_1 = BERTopic(umap_model=umap_model,
                       hdbscan_model=cluster_model,
                       vectorizer_model=vectorizer_model)
```

```python
# Incrementally fit the topic model by training on 1000 documents at a time
for doc in doc_chunks:
    topic_model_1.partial_fit(doc)
```

```python
from datasets import load_dataset

dataset = load_dataset("CShorten/ML-ArXiv-Papers")["train"]

# Extract abstracts to train on and corresponding titles
abstracts = dataset["abstract"]
```

```python
from bertopic import BERTopic
import transformers
from sentence_transformers import SentenceTransformer
from bertopic.representation import TextGeneration

# Load an embedding model (default setup with MiniLM)

sentence_model = SentenceTransformer("all-MiniLM-L6-v2")
embeddings = sentence_model.encode(abstracts, show_progress_bar=True)

# Initialize the text generation pipeline with an LLM
generator = transformers.pipeline("text2text-generation", model="google/flan-t5-base", device="cuda")

# Define a prompt for the LLM to summarize topics
prompt = system_prompt + example_prompt + main_prompt

# Create a TextGeneration representation model using the LLM
representation_model = {
    "FlanT5 - Base": TextGeneration(generator, prompt=prompt)
}

# Initialize BERTopic with the embedding model and LLM for enhanced topic representation
topic_model = BERTopic(embedding_model=sentence_model,
                       representation_model=representation_model,
                       # Hyperparameters
                       top_n_words=10,
                       verbose=True)

# Fit and transform documents to generate topics
topics, probs = topic_model.fit_transform(abstracts, embeddings)
```

```python
# Display the topic information with enhanced representations
topic_info = topic_model.get_topic_info()
labels = [item[0] for item in topic_info["FlanT5 - Base"]]
topic_model.set_topic_labels(labels)
topic_info = topic_model.get_topic_info()
topic_info
```
